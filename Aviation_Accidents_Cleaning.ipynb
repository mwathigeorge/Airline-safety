### Cleaning and constructing Key Measurables

Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# Select relevant columns
injury_damage_cols = [
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured',
    'Aircraft.damage'
]

# Display the first five records
df[injury_damage_cols].head()
```

```python
# Identify injury columns
injury_cols = [
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured'
]

# Convert injury columns to numeric
for col in injury_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Replace negative injury counts with NaN
for col in injury_cols:
    df.loc[df[col] < 0, col] = np.nan

# Check missing values after cleaning
df[injury_cols].isnull().sum()
```

```python
# Identify records where all injury counts are available
complete_injury_records = df[injury_cols].notna().all(axis=1)

# Create Total.Occupants
df['Total.Occupants'] = np.nan

df.loc[complete_injury_records, 'Total.Occupants'] = (
    df.loc[complete_injury_records, injury_cols].sum(axis=1)
)

# Display the results
df[[
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured',
    'Total.Occupants'
]].head(10)
```

```python
# Calculate combined fatal and serious injuries
df['Fatal.Serious.Injuries'] = np.nan

df.loc[complete_injury_records, 'Fatal.Serious.Injuries'] = (
    df.loc[complete_injury_records, 'Total.Fatal.Injuries'] +
    df.loc[complete_injury_records, 'Total.Serious.Injuries']
)

# Calculate fatal/serious injury rate
# Replace zero occupants with NaN to prevent division by zero

df['Fatal.Serious.Injury.Rate'] = (
    df['Fatal.Serious.Injuries'] /
    df['Total.Occupants'].replace(0, np.nan)
)

# Display the new variables
df[[
    'Total.Occupants',
    'Fatal.Serious.Injuries',
    'Fatal.Serious.Injury.Rate'
]].head(10)
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
# Clean the Aircraft.damage column

# Remove leading/trailing spaces and standardize capitalization
df['Aircraft.damage'] = (
    df['Aircraft.damage']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Standardize unknown values
df['Aircraft.damage'] = df['Aircraft.damage'].replace({
    'UNK': 'UNKNOWN',
    '': pd.NA
})

# Create a binary column to indicate whether the aircraft was destroyed
# 1 = Destroyed
# 0 = Not destroyed (Substantial or Minor damage)
# NaN = Missing or unknown damage information

df['Aircraft.Destroyed'] = df['Aircraft.damage'].map({
    'DESTROYED': 1,
    'SUBSTANTIAL': 0,
    'MINOR': 0
})

# Display the results
df[['Aircraft.damage', 'Aircraft.Destroyed']].head(10)

# Check the distribution of destruction outcomes
print(df['Aircraft.Destroyed'].value_counts(dropna=False))

# Calculate the aircraft destruction rate
destruction_rate = df['Aircraft.Destroyed'].mean() * 100

print(f"Aircraft destruction rate: {destruction_rate:.2f}%")
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
# 1. Remove leading and trailing spaces
df['Make'] = (
    df['Make']
    .astype('string')
    .str.strip()
    .str.upper()
    .str.replace(r'\s+', ' ', regex=True)
)

# 2. Replace inconsistent manufacturer names
make_aliases = {
    'AIR TRACTOR INC': 'AIR TRACTOR',
    'CIRRUS DESIGN CORP': 'CIRRUS',
    'CIRRUS DESIGN CORPORATION': 'CIRRUS',
    'DE HAVILLAND': 'DEHAVILLAND',
    'DE HAVILLAND AIRCRAFT': 'DEHAVILLAND',
    'AVIAT AIRCRAFT INC': 'AVIAT'
}

df['Make'] = df['Make'].replace(make_aliases)

# 3. Remove missing, blank and unknown manufacturer names
df = df.dropna(subset=['Make'])

df = df[
    ~df['Make'].isin(['', 'UNKNOWN', 'UNK'])
].copy()

# 4. Count accident records for each manufacturer
make_counts = df['Make'].value_counts()

# 5. Retain manufacturers with at least 50 accident records
eligible_makes = make_counts[make_counts >= 50].index

df = df[df['Make'].isin(eligible_makes)].copy()

# 6. Display the manufacturer counts
print("Number of eligible manufacturers:", df['Make'].nunique())

print("\nAccident records per manufacturer:")
print(df['Make'].value_counts())

# 7. Check the cleaned Make column
print("\nFirst 10 manufacturer names:")
print(df['Make'].head(10))

# 8. Check for missing values
print("\nMissing manufacturer names:", df['Make'].isna().sum())
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# 1. Remove missing model names
df = df.dropna(subset=['Model']).copy()

# 2. Remove leading/trailing spaces and standardize capitalization
df['Model'] = (
    df['Model']
    .astype('string')
    .str.strip()
    .str.upper()
    .str.replace(r'\s+', ' ', regex=True)
)

# 3. Remove blank and unknown model names
df = df[
    ~df['Model'].isin(['', 'UNKNOWN', 'UNK'])
].copy()

# 4. Inspect the most common aircraft models
print("Top 20 Aircraft Models:")
print(df['Model'].value_counts().head(20))

# 5. Count accident records for each Make and Model
make_model_counts = (
    df.groupby(['Make', 'Model'])
    .size()
    .reset_index(name='Accident_Count')
    .sort_values(by='Accident_Count', ascending=False)
)

print("\nAccident counts for each Make and Model:")
print(make_model_counts.head(20))

# 6. Check whether model names are unique to each manufacturer
model_make_counts = df.groupby('Model')['Make'].nunique()

shared_models = model_make_counts[model_make_counts > 1]

print("\nNumber of model labels associated with multiple manufacturers:")
print(len(shared_models))

print("\nExamples of model labels shared by different manufacturers:")
print(shared_models.head(20))

# 7. Create a unique aircraft identifier using Make and Model
df['Make_Model'] = df['Make'] + ' | ' + df['Model']

# 8. Inspect the new unique identifier
print("\nUnique aircraft identifiers:")
print(df['Make_Model'].value_counts().head(20))

# 9. Check the final results
print("\nTotal unique aircraft types:", df['Make_Model'].nunique())
print("Missing model names:", df['Model'].isna().sum())

# Display final dataframe
df[['Make', 'Model', 'Make_Model']].head(10)
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python

# 1. INSPECT ORIGINAL VALUES


columns_to_clean = [
    'Engine.Type',
    'Weather.Condition',
    'Number.of.Engines',
    'Purpose.of.flight',
    'Broad.phase.of.flight'
]

# Display data types and missing values
print("DATA TYPES:")
print(df[columns_to_clean].dtypes)

print("\nMISSING VALUES:")
print(df[columns_to_clean].isnull().sum())

# Inspect unique values in each column
for col in columns_to_clean:
    print(f"\nUnique values in {col}:")
    print(df[col].value_counts(dropna=False))


# Remove extra spaces and standardize capitalization
df['Engine.Type'] = (
    df['Engine.Type']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Standardize unknown values
df['Engine.Type'] = df['Engine.Type'].replace({
    'UNK': 'UNKNOWN',
    'NONE': 'UNKNOWN',
    '': pd.NA
})


# Remove extra spaces and standardize capitalization
df['Weather.Condition'] = (
    df['Weather.Condition']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Standardize unknown values
df['Weather.Condition'] = df['Weather.Condition'].replace({
    'UNK': 'UNKNOWN',
    '': pd.NA
})

# Convert number of engines to numeric
df['Number.of.Engines'] = pd.to_numeric(
    df['Number.of.Engines'],
    errors='coerce'
)

# Replace zero or negative engine counts with NaN
# Conventional airplanes must have at least one engine
df.loc[
    df['Number.of.Engines'] <= 0,
    'Number.of.Engines'
] = np.nan


# Remove extra spaces and standardize capitalization
df['Purpose.of.flight'] = (
    df['Purpose.of.flight']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Standardize unknown and blank values
df['Purpose.of.flight'] = df['Purpose.of.flight'].replace({
    'UNK': 'UNKNOWN',
    '': pd.NA
})

# Remove extra spaces and standardize capitalization
df['Broad.phase.of.flight'] = (
    df['Broad.phase.of.flight']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Standardize unknown and blank values
df['Broad.phase.of.flight'] = df['Broad.phase.of.flight'].replace({
    'UNK': 'UNKNOWN',
    '': pd.NA
})


# Display cleaned unique values and counts
for col in columns_to_clean:
    print(f"\nCleaned values for {col}:")
    print(df[col].value_counts(dropna=False))

# Check missing values after cleaning
print("\nMISSING VALUES AFTER CLEANING:")
print(df[columns_to_clean].isnull().sum())

# Display the cleaned columns
df[columns_to_clean].head(10)
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
# Inspect missing values in each column
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': (df.isnull().mean() * 100).round(2)
})

print("Missing values before column removal:")
display(missing_summary.sort_values(
    by='Missing Percentage', ascending=False
))

# Identify columns with more than 50% missing values
columns_to_drop = missing_summary[
    missing_summary['Missing Percentage'] > 50
].index

print("\nColumns to be removed:")
print(columns_to_drop.tolist())

# Drop columns with more than 50% missing values
df.drop(columns=columns_to_drop, inplace=True)

# Inspect the dataframe after removal
print("\nDataframe shape after column removal:", df.shape)
display(df.head())
```
